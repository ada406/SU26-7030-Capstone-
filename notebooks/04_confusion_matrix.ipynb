# Confusion matrix (test set)

Visualizes how often the model confuses each true genre with each predicted genre.

- **Raw counts:** number of test tracks
- **Normalized:** fraction within each true genre (rows sum to ~1)

Uses `outputs/test_predictions.csv` when available (otherwise re-evaluates the saved model).

### How to read it
- Rows = **true** genre, columns = **predicted** genre
- Dark diagonal cells = correct predictions
- Off-diagonal dark cells = common confusions (e.g. rock predicted as metal)
