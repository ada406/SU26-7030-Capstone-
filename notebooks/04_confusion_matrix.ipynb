# 04 — Confusion matrix (main result visualization)

Visualizes how often the model confuses each **true parent genre** with each **predicted** genre on the held-out test set.

- **Raw counts** (`outputs/confusion_matrix.png`): number of test tracks
- **Normalized** (`outputs/confusion_matrix_normalized.png`): fraction within each true genre (rows sum to ~1) — **this is the project's main result figure**

Uses `outputs/test_predictions.csv` when available (otherwise re-evaluates the saved model).  
**Canonical CLI:** `python scripts/plot_confusion_matrix.py`

Committed figures already match the ~59.1% / 9-class model; you can view them without re-running if predictions are present.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
elif not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = Path("..").resolve()

sys.path.insert(0, str(PROJECT_ROOT))
print("Project root:", PROJECT_ROOT)

from IPython.display import Image, Markdown, display
from src.plots import CONFUSION_MATRIX_NORM_PATH, CONFUSION_MATRIX_PATH, make_confusion_matrix_figures
from src.train import load_metrics
```

```python
metrics = load_metrics()
print("Test accuracy:", round(metrics["accuracy"], 4))
print("Macro F1:", round(metrics["macro_f1"], 4))
print("n_classes:", metrics["n_classes"])

result = make_confusion_matrix_figures(show=False)
print("Genres:", result["labels"])
result["raw_path"], result["normalized_path"]
```

## Main result figure (row-normalized)

A strong diagonal means the model has real signal. Off-diagonal mass among pop / rock / urban / world shows where audio features alone are ambiguous.

```python
display(Markdown("### Normalized (primary figure)"))
display(Image(filename=str(result["normalized_path"])))
display(Markdown("### Raw counts"))
display(Image(filename=str(result["raw_path"])))
print("Saved:", CONFUSION_MATRIX_NORM_PATH)
print("Saved:", CONFUSION_MATRIX_PATH)
```

### How to read it
- Rows = **true** parent genre, columns = **predicted** parent genre
- Dark diagonal cells = correct predictions
- Off-diagonal dark cells = common confusions (e.g. rock predicted as metal or pop)

These images are also shown in the Streamlit **Model performance** tab and embedded in the README Findings section.
